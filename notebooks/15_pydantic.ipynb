```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name: str
    age : int
```

```python
user = User(name="Ahtesham", age=23)
user
```

```python
### Field Validation
class Product(BaseModel):
    title:str = Field(max_length=10)
    des:str = Field(max_length=50)
    price:int = Field(gt=0, le=100)
```

```python
product = Product(title='Sumsung', des= "This is amazing mobile from samsung...", price=20)
product2 = Product(title='iPhone', des= "This is amazing mobile from Apple...", price=50)
```

```python
product
```

```python
class Order(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    product:Product
```

```python
order = Order(id = "ndkjf9i3j", product=product)
```

```python
order
```

```python
order.product.price
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: List[Product]
    comment: str = Field(
        default="No Comment"
    )
```

```python
cart = Carts(id="ak124m", items=[product, product2], comment="I don't have both phone")
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

```python
commonValidator = Annotated[List[Product], Field([])]
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: commonValidator
    comments: str = Field(
        default="No comment"
    )
```

```python
class Order(BaseModel):
    products: commonValidator
```

```python
ord = Order(products=[product, product2])
```
